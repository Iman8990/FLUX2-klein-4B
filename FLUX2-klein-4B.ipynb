```python
%%writefile /content/panel_app.py
# FLUX.2 [Klein] - 4B panel — Gradio UI driving a local headless ComfyUI server.
# Weights: black-forest-labs FP8 transformer (flux-2-klein-4b-fp8) +
# Comfy-Org qwen_3_4b encoder (shared with the Z-Image stack) + flux2 VAE.
# Graph transcribed from the official ComfyUI template
# image_flux2_klein_text_to_image.json (comfy-core 0.8.2 subgraph):
# UNETLoader -> CFGGuider(model, positive, negative, cfg) ; RandomNoise ;
# KSamplerSelect(euler) ; Flux2Scheduler(steps, width, height) ;
# EmptyFlux2LatentImage ; SamplerCustomAdvanced ; VAEDecode.
# Distilled operating point: 4 steps, CFG 1.0 (the template's Base
# subgraph ships 20 steps / CFG 5 for the base checkpoint).
import io
import json
import random
import time
import urllib.request
import urllib.parse
import uuid

import gradio as gr
import numpy as np
from PIL import Image

COMFY = "http://127.0.0.1:8188"
MAX_SEED = np.iinfo(np.int32).max
CLIENT_ID = uuid.uuid4().hex

RESOLUTIONS = [
    "1024x1024 (1:1)", "1152x896 (4:3-ish)", "896x1152 (3:4-ish)",
    "1280x720 (16:9)", "720x1280 (9:16)", "1344x768 (7:4)",
    "768x1344 (4:7)", "1536x640 (12:5)", "640x1536 (5:12)",
]

EXAMPLES = [
    "A portrait photo of a confident young woman in a leather jacket standing next to a vintage motorcycle, golden hour light, city street background, photorealistic, cinematic",
    "A hedgehog wearing a tiny party hat surrounded by confetti, early digital camera style, slight noise, flash photography, candid moment, 2000s digicam aesthetic, festive birthday celebration atmosphere",
    "A portrait photo of a young man with curly hair and glasses, wearing a denim jacket, golden hour light in a city park, shallow depth of field, photorealistic",
    "A cozy bookshop interior at night, warm lamps, wooden shelves full of old books, a cat sleeping on the counter, cinematic atmosphere",
]

CSS = """
.main-title { text-align: center; margin-bottom: 5px; }
.main-title h1 { font-weight: 750; margin-bottom: 3px; font-size: 2em; }
.main-title p { color: #6b6f76; margin: 0 auto; max-width: 700px; }
.credits-line { text-align: center; font-size: 0.9em; color: #868b94; margin-top: -2px; margin-bottom: 8px; }
#generate-btn { background: linear-gradient(135deg, #2563eb, #1d4ed8); color: white; border: none; }
"""

def comfy_ready():
    try:
        with urllib.request.urlopen(COMFY + "/system_stats", timeout=3) as r:
            return r.status == 200
    except Exception:
        return False

def post_prompt(workflow):
    data = json.dumps({"prompt": workflow, "client_id": CLIENT_ID}).encode()
    req = urllib.request.Request(COMFY + "/prompt", data=data,
                                 headers={"Content-Type": "application/json"})
    with urllib.request.urlopen(req, timeout=30) as r:
        return json.loads(r.read())["prompt_id"]

def wait_images(prompt_id, timeout=900):
    t0 = time.time()
    while time.time() - t0 < timeout:
        try:
            with urllib.request.urlopen(f"{COMFY}/history/{prompt_id}", timeout=10) as r:
                hist = json.loads(r.read())
        except Exception:
            time.sleep(2); continue
        entry = hist.get(prompt_id)
        if entry:
            status = entry.get("status", {})
            if status.get("status_str") == "error":
                msgs = status.get("messages", [])
                raise RuntimeError(f"ComfyUI error: {msgs[-1] if msgs else 'unknown'}")
            imgs = []
            for node_out in entry.get("outputs", {}).values():
                for im in node_out.get("images", []):
                    q = urllib.parse.urlencode(
                        {"filename": im["filename"], "subfolder": im.get("subfolder", ""),
                         "type": im.get("type", "output")})
                    with urllib.request.urlopen(f"{COMFY}/view?{q}", timeout=30) as r:
                        imgs.append(Image.open(io.BytesIO(r.read())).convert("RGB"))
            if imgs:
                return imgs
        time.sleep(2)
    raise TimeoutError("Generation timed out")

def parse_dims(label):
    wh = (label or "1024x1024").split("(")[0].strip()
    w, h = wh.split("x")
    return int(w), int(h)

def generate(prompt, steps, resolution, seed):
    if not comfy_ready():
        raise gr.Error("سرور ComfyUI هنوز بالا نیامده — پیام آماده‌شدن نوت‌بوک را ببین و دوباره بزن")
    prompt = (prompt or "").strip()
    if not prompt:
        raise gr.Error("پرامپت خالی است")
    if int(seed) < 0:
        seed = random.randint(0, MAX_SEED)
    w, h = parse_dims(resolution)
    nodes = {
        "1": {"class_type": "UNETLoader",
              "inputs": {"unet_name": "flux-2-klein-4b-fp8.safetensors", "weight_dtype": "default"}},
        "3": {"class_type": "CLIPLoader",
              "inputs": {"clip_name": "qwen_3_4b.safetensors", "type": "flux2", "device": "default"}},
        "6": {"class_type": "VAELoader", "inputs": {"vae_name": "flux2-vae.safetensors"}},
        "4": {"class_type": "CLIPTextEncode", "inputs": {"clip": ["3", 0], "text": prompt}},
        "5": {"class_type": "CLIPTextEncode", "inputs": {"clip": ["3", 0], "text": ""}},
        "63": {"class_type": "CFGGuider",
               "inputs": {"model": ["1", 0], "positive": ["4", 0], "negative": ["5", 0], "cfg": 1.0}},
        "73": {"class_type": "RandomNoise", "inputs": {"noise_seed": int(seed)}},
        "61": {"class_type": "KSamplerSelect", "inputs": {"sampler_name": "euler"}},
        "62": {"class_type": "Flux2Scheduler",
               "inputs": {"steps": int(steps), "width": w, "height": h}},
        "66": {"class_type": "EmptyFlux2LatentImage",
               "inputs": {"width": w, "height": h, "batch_size": 1}},
        "64": {"class_type": "SamplerCustomAdvanced",
               "inputs": {"noise": ["73", 0], "guider": ["63", 0], "sampler": ["61", 0],
                          "sigmas": ["62", 0], "latent_image": ["66", 0]}},
        "9": {"class_type": "VAEDecode", "inputs": {"samples": ["64", 0], "vae": ["6", 0]}},
        "11": {"class_type": "SaveImage",
               "inputs": {"images": ["9", 0], "filename_prefix": "flux2klein"}},
    }
    imgs = wait_images(post_prompt(nodes))
    return imgs[0], int(seed)

with gr.Blocks(title="FLUX.2 [Klein] - 4B") as demo:
    gr.HTML("""
    <div class="main-title">
        <h1>FLUX.2 [Klein] - 4B</h1>
        <p>Fast text-to-image model by Black Forest Labs.</p>
    </div>
    <p class="credits-line">FP8 · ComfyUI</p>
    """)
    with gr.Row():
        with gr.Column(scale=5):
            prompt = gr.Textbox(label="Prompt", lines=3,
                                placeholder="Describe the image you want to generate...")
            with gr.Row():
                steps = gr.Slider(label="Steps", minimum=1, maximum=20, value=4, step=1)
                seed = gr.Number(label="Seed", value=-1, precision=0)
            resolution = gr.Dropdown(label="Resolution", choices=RESOLUTIONS, value=RESOLUTIONS[0])
            gen_btn = gr.Button("Generate Image", elem_id="generate-btn")
        with gr.Column(scale=4):
            output = gr.Image(label="Generated Image", type="pil", height=460)
            used_seed = gr.Number(label="Seed used", precision=0)
    gr.Examples(examples=[[e] for e in EXAMPLES], inputs=[prompt], label="Example Prompts")
    gen_btn.click(fn=generate, inputs=[prompt, steps, resolution, seed], outputs=[output, used_seed])

demo.launch(share=True, css=CSS)
```
